# How do sales trend over time?

### Methodology

- Aggregate the sales per month per sub-category
- Plot the monthly sales

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_excel('sample_superstore.xls')
df
```

```python
df['order_date_month_no'] = df['Order Date'].dt.month
df
```

```python
df_filtered = df[
    (df['Order Date'].dt.year == 2026) &
    (df['Order Date'].dt.month <= 8)
]
df_filtered
```

```python
df_pivot = df_filtered.pivot_table(index='order_date_month_no', columns='Sub-Category', aggfunc='size')
df_pivot.loc['Total'] = df_pivot.sum()
df_pivot = df_pivot[df_pivot.loc['Total'].sort_values(ascending=False).index]
df_pivot = df_pivot.drop('Total')
df_pivot
```

```python
df_pivot.index = df_pivot.index.map(lambda x: pd.to_datetime(x, format='%m').strftime('%b'))
df_pivot.index.name = 'order_date_month'
df_pivot
```

```python
df_plot = df_pivot.iloc[:,:5]

sns.lineplot(data=df_plot, dashes=False, palette='muted')
sns.despine()
sns.set_style('white')

plt.title('Sales Trend across Top 5 Sub-Categories 2026')
plt.xlabel('2026')
plt.ylabel('Orders')
plt.legend().remove()

for i in range(5):
    plt.text(7.2, df_plot.iloc[-1, i], df_plot.columns[i])

plt.show()
```
