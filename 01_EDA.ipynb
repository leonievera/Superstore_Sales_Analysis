# Exploratory Data Analysis

- Amount of Orders in total
- Amount of Products in total which were ordered
- Amount of Customers in total who ordered
- Countries shipped to
- Cities with most orders
- Different Categories
- Different Sub-Categories with most orders

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_excel('sample_superstore.xls')
df
```

```python
orders_total = df['Order ID'].value_counts().count()
orders_total
```

```python
products_total = df['Product ID'].value_counts().count()
products_total
```

```python
customers_total = df['Customer ID'].value_counts().count()
customers_total
```

```python
countries_total = df['Country/Region'].value_counts()
countries_total
```

```python
plt.pie(countries_total, labels=['United States', 'Canada'], autopct='%1.1f%%', startangle=90)
plt.title('Distribution of the Countries')
plt.show()
```

```python
cities_total = df['City'].value_counts()
cities_total = cities_total.to_frame()
cities_orders = cities_total.head(10)
cities_orders
```

```python
sns.barplot(data=cities_orders, x='count', y='City')
sns.despine()
plt.title('Amount of Orders for top 10 Cities')
plt.xlabel('Orders')
plt.ylabel('')
plt.show()
```

```python
categories_total = df['Category'].value_counts()
categories_total
```

```python
sub_categories_total = df['Sub-Category'].value_counts()
sub_categories_total = sub_categories_total.to_frame()
sub_categories_orders = sub_categories_total.head(10)
sub_categories_orders
```

```python
sns.barplot(data=sub_categories_orders, x='count', y='Sub-Category')
sns.despine()
plt.title('Amount of Orders for top 10 Sub-Categories')
plt.xlabel('Orders')
plt.ylabel('')
plt.show()
```
