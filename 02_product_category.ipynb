# Which product category generates the most profit?

### Methodology

- Calculate profit based on categories
- Plot the findings

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_excel('sample_superstore.xls')
df
```

```python
df['order_date_month_no'] = df['Order Date'].dt.month
df
```

```python
df_filtered = df[
    (df['Order Date'].dt.year == 2026) &
    (df['Order Date'].dt.month <= 8)
]
df_filtered
```

```python
df_profit = df_filtered.groupby('Category')['Profit'].sum().sort_values(ascending=False)
df_profit = df_profit.to_frame()
df_profit
```

```python
sns.barplot(df_profit, x='Profit', y='Category')
sns.set_style('white')
sns.despine()
plt.title('Total Profit by Product Category for 2026')
plt.xlabel('Profit in USD')
plt.ylabel('')
ticks_x = plt.FuncFormatter(lambda x, pos: f'${int(x/1000):.0f}K')
plt.gca().xaxis.set_major_formatter(ticks_x)
plt.show()
```
